```python
plt.figure(figsize=(8,5))
sns.histplot(df['GPA'], bins=20, kde=True)

plt.title("Distribution of GPA")
plt.xlabel("GPA")
plt.ylabel("Number of Students")

plt.show()
```

Most students are concentrated in the lower-to-middle GPA ranges, with relatively fewer students achieving very high GPAs.

```python
plt.figure(figsize=(8,5))
sns.countplot(x='GradeClass', data=df)

plt.title("Grade Class Distribution")

plt.show()
```

GradeClass 4 contains the largest number of students, indicating a substantial population of lower-performing students.





```python
plt.figure(figsize=(8,5))
sns.histplot(df['Absences'], bins=20)

plt.title("Distribution of Student Absences")

plt.show()
```

Student absence rates vary considerably, suggesting attendance may be an important factor influencing academic performance.

```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='StudyTimeWeekly',
    y='GPA',
    data=df
)

plt.title("Study Time vs GPA")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='ParentalSupport',
    y='GPA',
    data=df
)

plt.title("Parental Support vs GPA")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    x='Tutoring',
    y='GPA',
    data=df
)

plt.title('GPA vs Tutoring')
plt.xlabel('Tutoring (0=No, 1=Yes)')
plt.ylabel('GPA')

plt.show()
```


Students who receive tutoring generally achieve higher GPAs than those who do not. This suggests that academic support programs may positively influence student performance.
